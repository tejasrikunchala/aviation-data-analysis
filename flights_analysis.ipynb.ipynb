## 15 Filling Missing Values in Delay Reason Columns

> NaN indicates that the flight was not cancelled due to that specific delay reason.
>
> We replaced missing values with 0 to represent no delay.

```python
# Delay columns - fill NaN with 0
delay_cols = ["delay_due_carrier","delay_due_weather","delay_due_nas","delay_due_security","delay_due_late_aircraft"  ]
df[delay_cols] = df[delay_cols].fillna(0)
```

```python

df[delay_cols].isnull().sum()
```

```python
df['cancellation_code']
```

## 16 Mapping Cancellation Reasons and Handling Missing Values

> A indicates Carrier Delay, B indicates Weather Delay,
>
> C indicates NAS Delay, and D indicates Security Delay.
>
> NaN indicates that the flight was not cancelled.We replaced NaN with "No Cancellation".



```python
##here nan represents that no cancellation
mapping = {
    "A": "Carrier",
    "B": "Weather",
    "C": "NAS",
    "D": "Security"
}
df['cancellation_code']=df['cancellation_code'].replace(mapping).fillna("no cancellation")
```

```python
df['cancellation_code']
```

```python
df["cancellation_code"].value_counts()
```

## 17 Cancellation Reason Count

> This analysis counts the number of cancelled flights for each cancellation reason by considering only flights.

```python
df[df['cancelled']==1]['cancellation_code'].value_counts()
```

For delay analysis, cancelled flights should not be included.
Why?

> Because cancelled flights never departed, so delay values become misleading.
>
> So we will create a new dataset with only operated flights.


> ## 18 Identifying Operated Flights
>
> Operated flights are flights that were not cancelled.
>
> We identified operated flights by filtering the Cancelled column
>
> where the value is 0.

```python
operated_flights=df[df['cancelled']==0]
```

```python
operated_flights.shape
```

Why This Step Is Important

Real airline analysts separate:

Operational flights → delay analysis

Cancelled flights → cancellation analysis

This is standard aviation analytics practice

## 19 Average Arrival Delay by Airline

> This analysis calculates the average arrival delay for each airline and sorts the results from highest to lowest.

```python
# Which airlines have the highest average arrival delay?

airline_delay=operated_flights.groupby('airline')['arr_delay'].mean().sort_values(ascending=False)

airline_delay
```

>The analysis shows that Allegiant Air, JetBlue, and Frontier Airlines have the highest average arrival delays, exceeding 10 minutes.
>
>In comparison, Delta Air Lines and Alaska Airlines have lower average arrival delays, close to one minute, indicating better average punctuality.

## 20 Top 10 Airports by Average Departure Delay


```python
#  Which airports have the highest average delays

airport_delay=operated_flights.groupby('origin')['dep_delay'].mean().sort_values(ascending=False)
airport_delay.head(10)
```

>These are the airports with fewer flights. Even a few delayed flights can increase the average delay.
>
> So, we should also check the total number of flights before deciding which airports have more delays.

## 21 Flights count by origin Analysis

> Analyzed airport traffic based on the number of flights

```python
# Which airports have the most flights?

origin_flight_count=operated_flights['origin'].value_counts()
origin_flight_count.head(10)
```

>The busiest airports are ATL, DFW, ORD, and DEN. These airports handle many flights,
>
> so delays can affect more passengers, airline schedules, and airport operations.
>
> Therefore, checking busy airports helps us understand the impact of flight delays more clearly.

## 22 Average Departure Delay at Busy Airports


```python
# Among the busiest airports, which ones have the highest delays?

busy_airports=origin_flight_count.head(10).index
operated_flights[operated_flights['origin'].isin(busy_airports)].groupby('origin')['dep_delay'].mean().sort_values(ascending=False)
```

>Among the 10 busiest airports, MCO has the highest average departure delay (14.54 minutes), followed by DEN (13.40 minutes) and DFW (12.64 minutes).
>
> SEA has the lowest average departure delay (7.14 minutes) among these airports.
>
>This comparison helps us understand departure delays at busy airports and identify airports that may need improvements in operational efficiency

## 23 Average Arrival Delay by Airlines

```python
# Plot Airline Delay Chart

plt.figure(figsize=(12,6))
c=sns.barplot(x=airline_delay.values, y=airline_delay.index)
c.bar_label(c.containers[0])
plt.title("Average Arrival Delay by Airlines")
plt.xlabel("Average Delay minutes")
plt.ylabel("Airlines")
plt.show()
```


What You Will See

A horizontal bar chart showing:

Airlines with highest delays at the top

Airlines with lowest delays at the bottom

This makes it much easier to understand than raw numbers.

Why This Visualization Matters

-->Which airlines are most delayed

--> Which airlines are most punctual

Airlines use these insights to improve operational performance.

Example report statement:

Low-cost carriers such as Allegiant Air, JetBlue Airways, and Frontier Airlines show the highest average arrival delays, exceeding 10 minutes. In contrast, Delta Air Lines and Alaska Airlines demonstrate strong punctuality with delays close to one minute.

## 24 Analysing delay causes

```python
# Now we analyze why flights get delayed.

delay_causes=operated_flights[['delay_due_carrier','delay_due_weather', 'delay_due_nas', 'delay_due_security','delay_due_late_aircraft']].sum().sort_values(ascending=False)
delay_causes
```

>Late aircraft arrivals and carrier-related issues contribute the most to total flight delay minutes.
>
> NAS delays also have a significant impact, while weather and security delays contribute fewer minutes.
>
 > This analysis helps us understand the main causes of flight delays.

## 25 Total Delay Minutes by Causes

```python
plt.figure(figsize=(10,6))
a=sns.barplot(x=delay_causes.values, y=delay_causes.index)
plt.bar_label(a.containers[0])
plt.title("Total Delay Minutes by Causes")
plt.xlabel("Total Delay Minutes")
plt.ylabel("Delay Causes")
plt.show()
```

>This chart shows the total delay minutes for each cause. Late aircraft arrivals and carrier-related issues contribute the most to flight delays.
>
> These delays can affect airline operations and cause further delays in other flights.
>
>Delay Causes:
>
>Late Aircraft: When a plane arrives late from its previous flight, the next flight may also be delayed. This is called delay propagation.
>
>Carrier Delays: Delays caused by crew issues, aircraft maintenance, and airline operational problems.

## 26 Monthly Operated Flights Analysis

```python
operated_flights['month']=operated_flights['fl_date'].dt.month
```

```python

# calculate average delay by month

monthly_delay=operated_flights.groupby('month')['arr_delay'].mean()

monthly_delay
```

>Flight delays show strong seasonal patterns.
>
> The highest delays occur during the summer travel season (June–August),
>
> likely due to increased passenger demand and weather-related disruptions.
>
> Delay levels decrease significantly in September before rising again during the winter holiday period in December.

## 27 Average Flight Arrival Delay by Month

```python
# Visualize Monthly Delay Trend

plt.figure(figsize=(10,6))

sns.lineplot(x=monthly_delay.index, y=monthly_delay.values, marker='o')

plt.title("Average Flight Arrival Delay by Month")
plt.xlabel("Month")
plt.ylabel("Average Delay (minutes)")

plt.show()
```


Visualization clearly shows the seasonal delay pattern:

June and July --> highest delays

September --> lowest delays

December --> delays increase again

This is exactly the type of trend analysis airlines perform for planning operations.

## 28 Route-wise Operated Flights and Arrival Delay Analysis

> Analyzed operated flights and average arrival delay by route.
>
> This helps identify routes with higher delays.

```python
# Analyze Flight Routes
# Which flight routes have the highest delays

# Create Route Column
operated_flights['route']=operated_flights['origin']+" → "+operated_flights['dest']

```

```python
route_delay=operated_flights.groupby('route')['arr_delay'].mean().sort_values(ascending=False)

route_delay.head(10)
# If a route had only 1 flight and that flight delayed 18 hours, the average becomes 1080 minutes.
```

## 29 Route count analysis

```python
# Filter Routes With Enough Flights

route_counts=operated_flights['route'].value_counts()
route_counts.head(10)
```


These are very high-traffic routes.

Example:

SFO ↔ LAX → California major airports

LAX ↔ LAS → Los Angeles – Las Vegas corridor

LGA ↔ ORD → New York – Chicago business route

These routes have thousands of flights, meaning they are major aviation corridors.

## 30 Filtering valid routes

```python
# routes with at least 100 flights.

valid_routes = route_counts[route_counts > 100].index
```

```python
# Now calculate delay again only for these routes

route_delay_filtered =operated_flights[operated_flights['route'].isin(valid_routes)].groupby('route')['arr_delay'] .mean().sort_values(ascending=False)
route_delay_filtered.head(10)
```

These routes show consistent delay patterns.

Possible reasons:

1.Airport congestion 2. Weather conditions 3. Distance / long-haul operations  Air traffic restrictions

Example:

ASE → ORD

ASE = Aspen (mountain airport) ORD = Chicago O'Hare (very busy hub)

Mountain airports often face weather and runway constraints, causing delays.

Why This Step Is Important

This removes statistical noise and shows real delay patterns.

Instead of rare routes, we analyze routes with significant traffic.

This is a very important skill recruiters look for:

Understanding when data can be misleading.
